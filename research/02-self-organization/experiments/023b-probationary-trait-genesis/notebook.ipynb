# Experiment 023b — Probationary Trait Genesis

Tests whether a temporary task-label-free fork can earn permanent existence through prospective counterfactual utility rather than hard cluster separability.


````python
from pathlib import Path
import os, subprocess, sys
ROOT = Path('/kaggle/working/mini-cells')
if not ROOT.exists():
    subprocess.run(['git','clone','https://github.com/ArcheLabs/mini-cells.git',str(ROOT)], check=True)
os.chdir(ROOT)
subprocess.run(['git','fetch','origin'], check=True)
subprocess.run(['git','switch','main'], check=True)
subprocess.run(['git','reset','--hard','origin/main'], check=True)
subprocess.run([sys.executable,'-m','pip','install','-q','-e','.[lm]'], check=True)
````

````python
tests = [
    'tests/research/02-self-organization/test_language_probationary_trait_genesis.py',
    'tests/research/02-self-organization/test_language_online_trait_genesis.py',
    'tests/research/02-self-organization/test_language_trait_bifurcation.py',
    'tests/research/02-self-organization/test_language_conflict_differentiation.py',
]
subprocess.run([sys.executable,'-m','pytest',*tests,'-q'], cwd=ROOT, check=True)
source_022 = ROOT/'artifacts'/'experiments'/'022-emergent-trait-bifurcation'/'decision.json'
source_023 = ROOT/'artifacts'/'experiments'/'023-online-nonparametric-trait-genesis'/'decision.json'
assert source_022.is_file(), 'merged Experiment 022 decision.json is required for provenance'
assert source_023.is_file(), 'merged Experiment 023 decision.json is required for provenance'
print('preflight passed')
````

````python
subprocess.run([sys.executable,'scripts/research/run_probationary_trait_genesis.py'], cwd=ROOT, check=True)
````
