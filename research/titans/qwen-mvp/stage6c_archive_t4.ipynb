```
from pathlib import Path
import subprocess
import sys
import tempfile

out = Path('/kaggle/working/stage6c-update64-archive')
out.mkdir(exist_ok=True)
with tempfile.TemporaryDirectory() as tmp:
    repo = Path(tmp) / 'repo'
    subprocess.run(['git', 'clone', '--depth=1', '--branch', 'research/minicells-08-runtime',
                    'https://github.com/ArcheLabs/mini-cells.git', str(repo)], check=True, timeout=300)
    subprocess.run([sys.executable, '-m', 'pip', 'install', 'transformers==5.3.0',
                    'datasets==4.4.1'], check=True, timeout=600)
    subprocess.run([sys.executable, '-u', str(repo / 'research/titans/qwen-mvp/archive64.py'),
                    str(out)], cwd=repo, check=True, timeout=2400)
```
