# MINI Cells Experiment 022 — Emergent Trait Bifurcation

Tests persistent K=2 structure in the phenotype-gradient field and compares geometry-driven bifurcation with an exactly stratified capacity control.


```python
from pathlib import Path
import os, subprocess, sys
ROOT = Path('/kaggle/working/mini-cells')
if not ROOT.exists():
    subprocess.run(['git','clone','https://github.com/ArcheLabs/mini-cells.git',str(ROOT)], check=True)
subprocess.run(['git','fetch','origin'], cwd=ROOT, check=True)
subprocess.run(['git','switch','main'], cwd=ROOT, check=True)
subprocess.run(['git','reset','--hard','origin/main'], cwd=ROOT, check=True)
subprocess.run([sys.executable,'-m','pip','install','-q','-e',str(ROOT)+'[lm]'], check=True)
print('repo:', ROOT)
```

## Preflight
022 is self-contained and does not require 019–021 model checkpoints. It reuses the validated TextNCA architecture and regenerates the deterministic dual-ability corpus.


```python
tests = [
    'tests/test_language_trait_bifurcation.py',
    'tests/test_language_conflict_differentiation.py',
]
subprocess.run([sys.executable,'-m','pytest',*tests,'-q'], cwd=ROOT, check=True)
```

## Run Experiment 022
Three replicates run on up to two GPUs. Parent and arm checkpoints are independently reusable after interruption.


```python
subprocess.run([sys.executable,'scripts/run_emergent_trait_bifurcation.py'], cwd=ROOT, check=True)
decision = ROOT/'results'/'emergent-trait-bifurcation-v1'/'decision.json'
print(decision.read_text())
```
