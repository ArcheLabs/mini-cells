# MINI Cells — Experiment 003C: Native Continual Learning

This experiment starts from the exact 003B solved Q8.8 artifact already committed to `main`. It asks whether production-shaped deterministic SIGN-SPSA can preserve Echo while acquiring a small disjoint adaptation rule.

The old domain excludes the reserved `??` prefix. In the adaptation domain, `??` marks a sequence whose first payload symbol must advance by one in `abcdefghijklmnopqrstuvwxyz0123456789`; all other tokens remain Echo. The key metric is `new_changed_accuracy`, not aggregate token accuracy.

The experiment compares global and block-512 native updates, new-only adaptation versus 50/50 replay, and keeps the production V1 batch size (4), margin (256), perturbation (4), and step (1). No production runtime code is modified.
