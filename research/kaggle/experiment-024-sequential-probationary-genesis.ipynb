# Experiment 024 — Sequential Probationary Genesis

Tests whether one persistent TextNCA organism can repeatedly reject or commit temporary task-label-free trait births and follow the preregistered `1 → 1 → 2 → 2 → 2 → 3` developmental trajectory.


````python
from pathlib import Path
import os, subprocess, sys
ROOT = Path('/kaggle/working/mini-cells')
if not ROOT.exists():
    subprocess.run(['git','clone','https://github.com/ArcheLabs/mini-cells.git',str(ROOT)], check=True)
os.chdir(ROOT)
subprocess.run(['git','fetch','origin'], check=True)
subprocess.run(['git','switch','main'], check=True)
subprocess.run(['git','reset','--hard','origin/main'], check=True)
subprocess.run([sys.executable,'-m','pip','install','-q','-e','.[lm]'], check=True)
````

````python
tests = [
    'tests/test_language_sequential_probationary_genesis.py',
    'tests/test_language_probationary_trait_genesis.py',
    'tests/test_language_online_trait_genesis.py',
    'tests/test_language_trait_bifurcation.py',
    'tests/test_language_conflict_differentiation.py',
]
subprocess.run([sys.executable,'-m','pytest',*tests,'-q'], cwd=ROOT, check=True)
import json
required = {
    '022-emergent-trait-bifurcation': 'EMERGENT_TRAIT_BIFURCATION_SIGNAL',
    '023-online-nonparametric-trait-genesis': 'NO_ONLINE_TRAIT_GENESIS',
    '023b-probationary-trait-genesis': 'PROBATIONARY_TRAIT_GENESIS_SIGNAL',
}
for name, status in required.items():
    path = ROOT/'artifacts'/'experiments'/name/'decision.json'
    assert path.is_file(), f'merged {name} decision.json is required for provenance'
    payload = json.loads(path.read_text())
    assert payload.get('status') == status, (name, payload.get('status'), status)
print('Experiment 024 preflight passed')
````

````python
subprocess.run([sys.executable,'scripts/run_sequential_probationary_genesis.py'], cwd=ROOT, check=True)
````

````python
decision = json.loads((ROOT/'results'/'sequential-probationary-genesis-v1'/'decision.json').read_text())
print(json.dumps(decision, indent=2))
````
