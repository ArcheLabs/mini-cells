# MINI Cells Experiment 021 — Conflict-Driven Differentiation

A story-trained TextNCA encounters arithmetic. We test whether unlabeled phenotype-gradient conflict can split one shared-genome cellular population into two functionally distinct descendants.


```python
from pathlib import Path
import os, subprocess, sys
ROOT = Path('/kaggle/working/mini-cells')
if not ROOT.exists():
    subprocess.run(['git','clone','https://github.com/ArcheLabs/mini-cells.git',str(ROOT)], check=True)
subprocess.run(['git','fetch','origin'], cwd=ROOT, check=True)
subprocess.run(['git','switch','main'], cwd=ROOT, check=True)
subprocess.run(['git','reset','--hard','origin/main'], cwd=ROOT, check=True)
subprocess.run([sys.executable,'-m','pip','install','-q','-e',str(ROOT)+'[lm]'], check=True)
print('repo:', ROOT)
```

## Preflight
021 is self-contained apart from the standard TinyStories cache/download. It does **not** require the 019/020 donor checkpoints and can run in a fresh Kaggle session.


```python
tests = [
    'tests/test_language_conflict_differentiation.py',
    'tests/test_language_recruitment_numerics.py',
]
subprocess.run([sys.executable,'-m','pytest',*tests,'-q'], cwd=ROOT, check=True)
```

## Run Experiment 021
Three replicates use up to two GPUs. Parent and arm checkpoints are saved independently, so rerunning resumes completed work.


```python
subprocess.run([sys.executable,'scripts/run_conflict_driven_differentiation.py'], cwd=ROOT, check=True)
decision = ROOT/'results'/'conflict-driven-differentiation-v1'/'decision.json'
print(decision.read_text())
```

## Publish curated results
Set `MINICELLS_PUBLISH=0` to skip GitHub publication. The 12 `.pt` parent/arm checkpoints remain Kaggle-local.


```python
if os.environ.get('MINICELLS_PUBLISH','1') != '0':
    subprocess.run([sys.executable,'scripts/publish_experiment_021_results.py','--push'], cwd=ROOT, check=True)
else:
    print('publication skipped')
```
