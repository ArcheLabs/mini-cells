# MINI Cells Experiment 017 — Localized Cellular Learning

Tests whether a frozen existing organism can divert a new skill into compact newborn tissue, retain old language behavior, and transfer the skill by grafting only that newborn tissue.

```python
import os, pathlib, shutil, subprocess, sys
REF = os.environ.get('MINICELLS_REF', 'main')
ROOT = pathlib.Path('/kaggle/working/mini-cells')
if ROOT.exists(): shutil.rmtree(ROOT)
subprocess.run(['git','clone','--depth','1','--branch',REF,'https://github.com/ArcheLabs/mini-cells.git',str(ROOT)], check=True)
os.chdir(ROOT)
subprocess.run([sys.executable,'-m','pip','install','-q','-e','.[lm,dev]'], check=True)
print('ref=', REF)
```
