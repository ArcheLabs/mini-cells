# MINI Cells Experiment 024b — Independent Third-Trait Challenge

Two-phase run: cross-replicate functional-independence screening, then a frozen selected-capability weak/strong 2→3 probation challenge.

```python
from pathlib import Path
import json, os, subprocess, sys
ROOT = Path('/kaggle/working/mini-cells')
if not ROOT.exists():
    subprocess.run(['git','clone','https://github.com/ArcheLabs/mini-cells.git',str(ROOT)], check=True)
subprocess.run(['git','fetch','origin'], cwd=ROOT, check=True)
subprocess.run(['git','switch','main'], cwd=ROOT, check=True)
subprocess.run(['git','reset','--hard','origin/main'], cwd=ROOT, check=True)
subprocess.run([sys.executable,'-m','pip','install','-q','-e','.[lm]'], cwd=ROOT, check=True)
```

```python
tests = [
    'tests/test_language_independent_third_trait.py',
    'tests/test_language_sequential_probationary_genesis.py',
    'tests/test_language_probationary_trait_genesis.py',
    'tests/test_language_online_trait_genesis.py',
    'tests/test_language_trait_bifurcation.py',
    'tests/test_language_conflict_differentiation.py',
]
subprocess.run([sys.executable,'-m','pytest',*tests,'-q'], cwd=ROOT, check=True)
required = {
    '022-emergent-trait-bifurcation': 'EMERGENT_TRAIT_BIFURCATION_SIGNAL',
    '023-online-nonparametric-trait-genesis': 'NO_ONLINE_TRAIT_GENESIS',
    '023b-probationary-trait-genesis': 'PROBATIONARY_TRAIT_GENESIS_SIGNAL',
    '024-sequential-probationary-genesis': 'FIRST_BIRTH_WITHOUT_SECOND_TRAIT_GENESIS',
}
for experiment, expected in required.items():
    path = ROOT/'artifacts'/'experiments'/experiment/'decision.json'
    payload = json.loads(path.read_text())
    assert payload['status'] == expected, (experiment, payload['status'])
assert (ROOT/'artifacts'/'experiments'/'024-sequential-probationary-genesis'/'DIAGNOSIS.md').is_file()
```

```python
subprocess.run([sys.executable,'scripts/run_independent_third_trait.py'], cwd=ROOT, check=True)
```

```python
decision = json.loads((ROOT/'results'/'independent-third-trait-v1'/'decision.json').read_text())
print(json.dumps(decision, indent=2, sort_keys=True))
```
