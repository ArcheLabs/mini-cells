```python
import subprocess
import sys
from pathlib import Path

# The exact fetched source SHA is recorded automatically. No training.
SOURCE_REF = "codex/epclm001-incremental"
ROOT = Path("/kaggle/working/epclm001-code")
OUTPUT = Path("/kaggle/working/epclm001-output")


def run(*args):
    subprocess.run([str(a) for a in args], check=True)


if not (ROOT / ".git").exists():
    run(
        "git",
        "clone",
        "--filter=blob:none",
        "--no-checkout",
        "https://github.com/ArcheLabs/mini-cells.git",
        ROOT,
    )
run("git", "-C", ROOT, "fetch", "--depth=1", "origin", SOURCE_REF)
run("git", "-C", ROOT, "sparse-checkout", "init", "--no-cone")
run(
    "git",
    "-C",
    ROOT,
    "sparse-checkout",
    "set",
    "--no-cone",
    "/research/projects/cellular-language-models/",
    "/research/projects/continual-learning/context/",
    "/src/",
    "/shared/",
    "/pyproject.toml",
)
run("git", "-C", ROOT, "checkout", "--detach", "FETCH_HEAD")
entry = (
    ROOT
    / "research/projects/cellular-language-models/experiments/epclm-001-incremental-execution/run.py"
)
try:
    run(sys.executable, entry, "engineering", "--output", OUTPUT, "--publish-github")
finally:
    print("Save Version retains EPCLM-001 reports:", OUTPUT)
    print("No model training. Scientific H1/H2/H3 remain independently gated.")
```
