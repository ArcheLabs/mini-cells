# MINI Cells — Experiment 004: Tiny Arithmetic Capability

This milestone moves beyond Echo and asks whether the same 4,476-parameter cellular model can answer single-digit addition and subtraction questions.

The experiment uses 110 deterministic problems: `a plus b` where `a+b<=9`, and `a minus b` where `a>=b`. Expressions are encoded as `3plus4?` or `9minus3?`; only the final `?` cell is supervised with the answer digit. A fixed 80/20 split distinguishes memorization from held-out generalization.

It runs three bridges: conventional FP32 acquisition, exact Q8.8 post-training quantization, and native Guarded Block SPSA continual acquisition from the solved 003B Echo artifact. Final reports include CSV evidence plus learning curves, retention/capability scatter, operation heatmaps, and a capability summary chart.
