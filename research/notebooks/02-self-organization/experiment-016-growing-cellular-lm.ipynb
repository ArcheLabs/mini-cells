# MINI Cells Experiment 016 — Growing Cellular Language Model

End-to-end small-model test of a parameter-matched Transformer, a fixed cellular LM, and a growing cellular organism with utility-driven connect/prune, pressure-conflict-driven fork, frozen-genome local learning, and same-genome tissue transplantation.

```python
import os, pathlib, shutil, subprocess, sys
REF = os.environ.get('MINICELLS_REF', 'main')
ROOT = pathlib.Path('/kaggle/working/mini-cells')
if ROOT.exists(): shutil.rmtree(ROOT)
subprocess.run(['git','clone','--depth','1','--branch',REF,'https://github.com/ArcheLabs/mini-cells.git',str(ROOT)], check=True)
os.chdir(ROOT)
subprocess.run([sys.executable,'-m','pip','install','-q','-e','.[lm,dev]'], check=True)
print('ref=', REF)
```

```python
import torch, platform
print({'python': platform.python_version(), 'torch': torch.__version__, 'cuda': torch.version.cuda, 'gpu_count': torch.cuda.device_count(), 'gpus': [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())]})
assert torch.cuda.is_available()
```
