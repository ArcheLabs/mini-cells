# MINI Cells Experiment 023 — Online Nonparametric Trait Genesis

Unknown-K online trait genesis from an unlabeled phenotype-gradient field with an explicit structural complexity cost.


```python
from pathlib import Path
import os, subprocess, sys
ROOT = Path('/kaggle/working/mini-cells')
if not ROOT.exists():
    subprocess.run(['git','clone','https://github.com/ArcheLabs/mini-cells.git',str(ROOT)], check=True)
subprocess.run(['git','fetch','origin'], cwd=ROOT, check=True)
subprocess.run(['git','switch','main'], cwd=ROOT, check=True)
subprocess.run(['git','reset','--hard','origin/main'], cwd=ROOT, check=True)
subprocess.run([sys.executable,'-m','pip','install','-q','-e',str(ROOT)+'[lm]'], check=True)
print('repo:', ROOT)
```

## Preflight
023 is self-contained and does not require 019–022 model checkpoints. It requires the merged Experiment 022 curated decision for provenance.


```python
tests = [
    'tests/research/02-self-organization/test_language_online_trait_genesis.py',
    'tests/research/02-self-organization/test_language_trait_bifurcation.py',
    'tests/research/02-self-organization/test_language_conflict_differentiation.py',
]
subprocess.run([sys.executable,'-m','pytest',*tests,'-q'], cwd=ROOT, check=True)
source_022 = ROOT/'artifacts'/'experiments'/'022-emergent-trait-bifurcation'/'decision.json'
assert source_022.is_file(), 'merged Experiment 022 decision.json is required for provenance'
print('source 022:', source_022)
```

## Run Experiment 023
Three replicates run on up to two GPUs. Parent and developmental-stage checkpoints are independently reusable; reruns resume from the latest completed stage.


```python
subprocess.run([sys.executable,'scripts/research/run_online_nonparametric_trait_genesis.py'], cwd=ROOT, check=True)
decision = ROOT/'results'/'online-nonparametric-trait-genesis-v1'/'decision.json'
print(decision.read_text())
```

## Publish curated results
Set `MINICELLS_PUBLISH=0` to skip GitHub publication. The 15 `.pt` parent/stage checkpoints remain Kaggle-local.


```python
if os.environ.get('MINICELLS_PUBLISH','1') != '0':
    subprocess.run([sys.executable,'scripts/research/publish_experiment_023_results.py','--push'], cwd=ROOT, check=True)
else:
    print('publication skipped')
```
