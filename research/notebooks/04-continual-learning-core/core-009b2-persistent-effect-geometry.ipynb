# Core Validation 009B-2 — Persistent Effect Geometry

Run discovery first. Confirmation is forbidden until discovery publishes a committed `basis-lock.json`. Requires Kaggle GPU + Internet and `GITHUB_TOKEN`; `HF_TOKEN` is optional but recommended.

```python
import json, os, pathlib, shutil, subprocess
ROOT = pathlib.Path('/kaggle/working/mini-cells')
BRANCH = 'codex/core-validation-009b2-persistent-effect-geometry'
try:
    from kaggle_secrets import UserSecretsClient
    token = UserSecretsClient().get_secret('HF_TOKEN')
    if token: os.environ['HF_TOKEN'] = token
except Exception:
    pass
shutil.rmtree(ROOT, ignore_errors=True)
subprocess.run(['git','clone','--depth','1','--branch',BRANCH,'https://github.com/ArcheLabs/mini-cells.git',str(ROOT)], check=True)
subprocess.run(['python','-m','pip','install','-q','-e','.[lm]'], cwd=ROOT, check=True)
subprocess.run(['python','scripts/research/orchestrate_core_validation_009b2.py','--phase','discovery','--branch',BRANCH,'--device','cuda','--push-results'], cwd=ROOT, check=True)
decision = json.loads((ROOT / 'results/core-validation-009b2-persistent-effect-geometry/discovery/decision.json').read_text())
print(json.dumps(decision, indent=2, sort_keys=True))
print((ROOT / 'results/core-validation-009b2-persistent-effect-geometry/discovery/RESULTS.md').read_text())
assert decision.get('confirmation_allowed') is True, 'Discovery did not find a <=32D compact effect subspace. Stop; do not run confirmation.'
```

## Confirmation

Run this cell only after discovery completed with `confirmation_allowed=true`. It re-clones the branch so the committed basis lock is guaranteed to be present.

```python
import json, os, pathlib, shutil, subprocess
ROOT = pathlib.Path('/kaggle/working/mini-cells')
BRANCH = 'codex/core-validation-009b2-persistent-effect-geometry'
try:
    from kaggle_secrets import UserSecretsClient
    token = UserSecretsClient().get_secret('HF_TOKEN')
    if token: os.environ['HF_TOKEN'] = token
except Exception:
    pass
shutil.rmtree(ROOT, ignore_errors=True)
subprocess.run(['git','clone','--depth','1','--branch',BRANCH,'https://github.com/ArcheLabs/mini-cells.git',str(ROOT)], check=True)
lock = ROOT / 'research/validations/core-009b2-persistent-effect-geometry/basis-lock.json'
assert lock.is_file(), 'Committed basis-lock.json is missing; refresh after discovery publication.'
subprocess.run(['python','-m','pip','install','-q','-e','.[lm]'], cwd=ROOT, check=True)
subprocess.run(['python','scripts/research/orchestrate_core_validation_009b2.py','--phase','confirmation','--branch',BRANCH,'--device','cuda','--push-results'], cwd=ROOT, check=True)
decision = json.loads((ROOT / 'results/core-validation-009b2-persistent-effect-geometry/confirmation/decision.json').read_text())
print(json.dumps(decision, indent=2, sort_keys=True))
print((ROOT / 'results/core-validation-009b2-persistent-effect-geometry/confirmation/RESULTS.md').read_text())
```
