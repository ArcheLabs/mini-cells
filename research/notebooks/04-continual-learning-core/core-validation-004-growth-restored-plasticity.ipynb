# Core Validation 004 — Growth-Restored Plasticity
Formal Kaggle run for the final controlled CLM core-loop mechanism test. Smoke is execution-only; only the frozen three-seed CUDA run can emit a scientific decision.


```python
import os, shutil, subprocess, sys
from pathlib import Path
REPO='https://github.com/ArcheLabs/mini-cells.git'
BRANCH='codex/core-validation-004-growth-restored-plasticity'
ROOT=Path('/kaggle/working/mini-cells')
if ROOT.exists(): shutil.rmtree(ROOT)
subprocess.run(['git','clone',REPO,str(ROOT)],check=True)
subprocess.run(['git','checkout',BRANCH],cwd=ROOT,check=True)
os.chdir(ROOT)
print(subprocess.check_output(['git','rev-parse','HEAD'],text=True).strip())
```

```python
subprocess.run([sys.executable,'-m','pip','install','-e','.[dev]'],check=True)
```

```python
subprocess.run([sys.executable,'-m','pytest','-q','tests/research/04-continual-learning-core/test_core_validation_004.py'],check=True)
```

```python
SMOKE=ROOT/'results/core-validation-004-smoke'
subprocess.run([sys.executable,'scripts/research/run_core_validation_004.py','--smoke','--device','cpu','--out',str(SMOKE)],check=True)
subprocess.run([sys.executable,'scripts/research/report_core_validation_004.py','--out',str(SMOKE)],check=True)
import json
smoke=json.loads((SMOKE/'decision.json').read_text())
assert smoke['status']=='SMOKE_ONLY' and smoke['scientific_decision'] is False
smoke
```

```python
import torch
print({'torch':torch.__version__,'cuda':torch.version.cuda,'available':torch.cuda.is_available(),'gpu':torch.cuda.get_device_name(0) if torch.cuda.is_available() else None})
assert torch.cuda.is_available(), 'Formal Core Validation 004 requires CUDA'
```

```python
OUT=ROOT/'results/core-validation-004-growth-restored-plasticity'
subprocess.run([sys.executable,'scripts/research/run_core_validation_004.py','--device','cuda','--out',str(OUT)],check=True)
subprocess.run([sys.executable,'scripts/research/report_core_validation_004.py','--out',str(OUT)],check=True)
```

```python
import pandas as pd
from IPython.display import display, Image
decision=json.loads((OUT/'decision.json').read_text())
display(pd.read_csv(OUT/'gate-summary.csv'))
display(pd.read_csv(OUT/'seed-summary.csv'))
print(json.dumps(decision,indent=2))
display(Image(filename=str(OUT/'stability-plasticity-frontier.png')))
display(Image(filename=str(OUT/'growth-recovery.png')))
display(Image(filename=str(OUT/'plasticity-recovery-by-seed.png')))
```
