# Core Validation 009D — Compositional Operator Geometry

Run discovery first. Confirmation is forbidden until discovery publishes a committed `representation-lock.json`. Requires Kaggle GPU + Internet and `GITHUB_TOKEN`; `HF_TOKEN` is optional but recommended.

A discovery result of `OPERATOR_FACTOR_COMPRESSION_ONLY` is an intentional stop: rank-1 operator compression alone is not cross-write reuse.

```python
import json, os, pathlib, shutil, subprocess
ROOT = pathlib.Path('/kaggle/working/mini-cells')
BRANCH = 'codex/core-validation-009d-compositional-operator-geometry'
try:
    from kaggle_secrets import UserSecretsClient
    for name in ('GITHUB_TOKEN', 'HF_TOKEN'):
        try:
            token = UserSecretsClient().get_secret(name)
            if token: os.environ[name] = token
        except Exception:
            pass
except Exception:
    pass
shutil.rmtree(ROOT, ignore_errors=True)
subprocess.run(['git','clone','--depth','1','--branch',BRANCH,'https://github.com/ArcheLabs/mini-cells.git',str(ROOT)], check=True)
subprocess.run(['python','-m','pip','install','-q','-e','.[lm]'], cwd=ROOT, check=True)
subprocess.run(['python','scripts/research/orchestrate_core_validation_009d.py','--phase','discovery','--branch',BRANCH,'--device','cuda','--push-results'], cwd=ROOT, check=True)
decision = json.loads((ROOT / 'results/core-validation-009d-compositional-operator-geometry/discovery/decision.json').read_text())
print(json.dumps(decision, indent=2, sort_keys=True))
print((ROOT / 'results/core-validation-009d-compositional-operator-geometry/discovery/RESULTS.md').read_text())
assert decision.get('confirmation_allowed') is True, '009D did not find reusable compositional operator geometry. Stop; do not run confirmation.'
```

## Confirmation

Run only after discovery completed with `confirmation_allowed=true`. Re-cloning is mandatory so the committed representation lock is present.

```python
import json, os, pathlib, shutil, subprocess
ROOT = pathlib.Path('/kaggle/working/mini-cells')
BRANCH = 'codex/core-validation-009d-compositional-operator-geometry'
try:
    from kaggle_secrets import UserSecretsClient
    for name in ('GITHUB_TOKEN', 'HF_TOKEN'):
        try:
            token = UserSecretsClient().get_secret(name)
            if token: os.environ[name] = token
        except Exception:
            pass
except Exception:
    pass
shutil.rmtree(ROOT, ignore_errors=True)
subprocess.run(['git','clone','--depth','1','--branch',BRANCH,'https://github.com/ArcheLabs/mini-cells.git',str(ROOT)], check=True)
lock = ROOT / 'research/validations/core-009d-compositional-operator-geometry/representation-lock.json'
assert lock.is_file(), 'Committed representation-lock.json is missing; refresh after discovery publication.'
subprocess.run(['python','-m','pip','install','-q','-e','.[lm]'], cwd=ROOT, check=True)
subprocess.run(['python','scripts/research/orchestrate_core_validation_009d.py','--phase','confirmation','--branch',BRANCH,'--device','cuda','--push-results'], cwd=ROOT, check=True)
decision = json.loads((ROOT / 'results/core-validation-009d-compositional-operator-geometry/confirmation/decision.json').read_text())
print(json.dumps(decision, indent=2, sort_keys=True))
print((ROOT / 'results/core-validation-009d-compositional-operator-geometry/confirmation/RESULTS.md').read_text())
```
