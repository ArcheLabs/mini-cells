# MiniCells HybridCLM quickstart
This launcher calls the public API. Pin an immutable model revision before running.

```python
from minicells import CellPlacement, HybridCLM
MODEL = 'ibm-granite/granite-3.1-1b-a400m-base'
REVISION = '<immutable-hub-commit>'
hybrid = HybridCLM.from_pretrained(MODEL, revision=REVISION)
print(hybrid.inspect().to_dict())
hybrid.cellularize(CellPlacement(layer=7, experts='all'))
```
